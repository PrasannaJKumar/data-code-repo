Task6: Create a pyspark dataframe df1 reading the data from the above file using pyspark magic command %python

```python
%python
file_path = "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"

df1 = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(file_path)
)

df1.printSchema()
display(df1)
```

Task6: Create a pyspark dataframe df1 reading the data from the above file using pyspark magic command %python

```python
%python
df1.printSchema()
display(df1)
```

Task7: Write the above dataframe df1 data into a databricks table called 'default.mobile_os_usage' using pyspark magic command %python


```python
%python
df1 = (
    df1
    .withColumnRenamed(
        "Mobile Operating System",
        "mobile_operating_system"
    )
    .withColumnRenamed(
        "Percent of Usage",
        "percent_of_usage"
    )
)

df1.printSchema()
display(df1)
```

Task8: Write sql query to display the data loaded into the table 'default.mobile_os_usage' using the pyspark magic command %python


```python
%python
df1.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.mobile_os_usage")

print("Table created successfully")
```

Task8: Write sql query to display the data loaded into the table 'default.mobile_os_usage' using the pyspark magic command %python


```python
%sql
SELECT *
FROM workspace.default.mobile_os_usage;
```

```python
%python
result_df = spark.sql(
    "SELECT * FROM workspace.default.mobile_os_usage"
)

display(result_df)
print("Total records:", result_df.count())
```
