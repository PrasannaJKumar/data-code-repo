```python
%python

# Create dropdown widget for environment
dbutils.widgets.dropdown(
    "environment",
    "dev",
    ["dev", "qa", "prod"],
    "Select Environment"
)

# Create text widget for owner name
dbutils.widgets.text(
    "owner_name",
    "prasanna",
    "Enter Owner Name"
)

print("Widgets created successfully!")

environment = dbutils.widgets.get("environment")
owner_name = dbutils.widgets.get("owner_name")

print("Environment:", environment)
print("Owner Name:", owner_name)
     
```

dbutils widget Usecase – Read Widget Values environment and owner and print in the screen

```python
%python

# Read widget values
environment = dbutils.widgets.get("environment")
owner_name = dbutils.widgets.get("owner_name")

# Print widget values
print("Selected Environment:", environment)
print("Owner Name:", owner_name)
     
```

dbutils widget Usecase – Move the above processed File to Archive

```python

%python

# Define source and destination paths
source_file = f"{processed_path}/patient_data.csv"
destination_file = f"{archive_path}/patient_data.csv"

# Move file from processed to archive
dbutils.fs.mv(source_file, destination_file)

print("File moved successfully from processed to archive!")

print("Processed folder:")
display(dbutils.fs.ls(processed_path))

print("Archive folder:")
display(dbutils.fs.ls(archive_path))
```
