```python
dbutils.widgets.dropdown("dropdown_widget","Prasanna",["Prasanna","Mani","Arun"],"Select your name")
aspirant_name_chosen=dbutils.widgets.get("dropdown_widget")
print("Good morning",aspirant_name_chosen)
```

```python

dbutils.widgets.multiselect(
    "multiselect_widget",
    "wd36",
    ["wd32", "we43", "we45", "wd36", "wd38", "we49"],
    "Select your team name"
)

all_batches = dbutils.widgets.get("multiselect_widget")

print(all_batches)
print(type(all_batches))

all_batches_list = all_batches.split(",")
print(all_batches_list)
print(type(all_batches_list))

if len(all_batches_list) > 0:
    batch1 = all_batches_list[0]
    print("Batch 1:", batch1)

if len(all_batches_list) > 1:
    batch2 = all_batches_list[1]
    print("Batch 2:", batch2)

if len(all_batches_list) > 2:
    batch3 = all_batches_list[2]
    print("Batch 3:", batch3)
```
