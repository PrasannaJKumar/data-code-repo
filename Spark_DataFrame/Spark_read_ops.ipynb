To provide schema (columname & datatype), what are the 2 basic options available that we learned so far ? inferSchema/toDF
Inferschema Benifit - Auto schema, drawbacks - takes more time to read the data and also it will not work for custom data types.
We are going to learn additionally 2 more options to handle schema (colname & datatype)?

Using simple string format of define schema (non familiar way, but simple way).
IMPORTANT: Using structure type to define schema. (familiar and market standard way, where we can go with more options of using complex schema also) acts as an alternative for inferschema - to define SIMPLE/COMPLEX custom schema & to improve performance also


```python
#By default it will use _c0,_c1..._cn it will apply as column headers, if we use toDF(colnames) we can define our own headers..

csv_df1=spark.read.csv("/Volumes/catalog1_dropme/schema1_dropme/commondata/custs").toDF("id","fname","lname","age","prof")
print(csv_df1.printSchema())
csv_df1=spark.read.csv("/Volumes/catalog1_dropme/schema1_dropme/commondata/custs",inferSchema=True).toDF("id","fname","lname","age","prof")
print(csv_df1.printSchema())

#1. Using simple string format of define custom simple schema.
str_struct="id integer,fname string,lname string,age integer,prof string"
csv_df1=spark.read.schema(str_struct).csv("/Volumes/catalog1_dropme/schema1_dropme/commondata/custs")
print(csv_df1.printSchema())
csv_df1.show(2)

#2. Important part - Using structure type to define custom complex schema.
#4000001,Kristina,Chung,55,Pilot
#pattern - 
#import the types library based classes..
# define_structure=StructType([StructField("colname",DataType(),True),StructField("colname",DataType(),True)...])
from pyspark.sql.types import StructType,StructField,StringType,IntegerType
custom_schema=StructType([StructField("id",IntegerType(),False),StructField("fname",StringType(),True),StructField("lname",StringType(),True),StructField("age",IntegerType(),True),StructField("prof",StringType())])
csv_df1=spark.read.schema(custom_schema).csv("/Volumes/catalog1_dropme/schema1_dropme/commondata/custs")
print(csv_df1.printSchema())
csv_df1.show(2)
     
```
